```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
from skimage import io
```

Menampilkan Gambar dengan OpenCV

```python
img = cv2.imread('/content/drive/MyDrive/PCVK/Ktm.jpg')

cv2_imshow(img)
```

Menampilkan Gambar dengan Matplotlib

```python
plt.imshow(img)
plt.show()
```

RGB

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.show()
```

Menampilkan Grayscale dan Magma

```python
img_gray = cv2.imread(
    '/content/drive/MyDrive/PCVK/Ktm.jpg',
    cv2.IMREAD_GRAYSCALE
)

plt.imshow(img_gray, cmap='gray')
plt.show()
plt.imshow(img_gray, cmap='magma')
plt.show()
```

Resize Gambar

```python
img_resize = cv2.resize(img_rgb, (1000, 1500))

plt.imshow(img_resize)
plt.show()
```

Flipping / Membalik Gambar

```python
img_flip = cv2.flip(img_resize, 1)

plt.imshow(img_flip)
plt.show()
```

```python
img_flip_vertical = cv2.flip(img_resize, 0)

plt.imshow(img_flip_vertical)
plt.show()
```

Membuat Black Image

```python
black_img = np.zeros((512, 512, 3), dtype=np.int16)

cv2_imshow(black_img)
```

Membuat Rectangle

```python
cv2.rectangle(
    black_img,
    pt1=(350, 50),
    pt2=(500, 150),
    color=(0, 255, 0),
    thickness=15
)

cv2_imshow(black_img)
```

Membuat Rectangle Kedua

```python
cv2.rectangle(
    black_img,
    pt1=(200, 200),
    pt2=(300, 300),
    color=(255, 0, 0),
    thickness=15
)

cv2_imshow(black_img)
```

Membuat Circle

```python
cv2.circle(
    black_img,
    center=(100, 100),
    radius=50,
    color=(0, 0, 255),
    thickness=8
)

cv2_imshow(black_img)
```

```python
Membuat Garis Diagonal
```

```python
cv2.line(
    black_img,
    pt1=(0, 0),
    pt2=(512, 512),
    color=(255, 255, 255),
    thickness=5
)

cv2_imshow(black_img)
```

Menambahkan Tulisan

```python
font = cv2.FONT_HERSHEY_SIMPLEX

cv2.putText(
    black_img,
    'Hello',
    (50, 480),
    font,
    1.5,
    (255, 255, 0),
    2,
    cv2.LINE_AA
)

cv2_imshow(black_img)
```

Membuat Polyline

```python
black_img2 = np.zeros(
    shape=(512, 512, 3),
    dtype=np.int32
)

cv2_imshow(black_img2)
```

```python
vertices = np.array(
    [[100,300],
     [200,200],
     [400,300],
     [200,400]],
    dtype=np.int32
)

vertices
```

```python
pts = vertices.reshape((-1,1,2))
```

```python
cv2.polylines(
    black_img2,
    [pts],
    isClosed=True,
    color=(0, 255, 255),
    thickness=5
)

cv2_imshow(black_img2)
```
